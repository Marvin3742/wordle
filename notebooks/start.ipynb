### Clean word data set. (Keep only 5 letter words)

```python
import pandas as pd
```

```python
WORDS_ALPHA = "../data/words_alpha.txt"
WORDS = "../data/words.txt"
with open(WORDS_ALPHA, "r") as f:
    lines = f.readlines()

with open(WORDS, "w") as f:
    header = "Words: \n"
    f.write(header)
    for line in lines:
        if len(line) == 6:
            f.write(line)
```
